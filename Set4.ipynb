```python
#Ask the user to enter a year and use if-else to check if it is a leap year.
year = int(input("Enter the Year = "))
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
  print("The year is a leap year")
else:
  print("The year is not a leap year")
```

```python
 #Write a program to find the roots of a quadratic equation using if-else
import math
import cmath
print("Enter the coefficients of ax^2 + bx + c = 0")
a = float(input("Enter the value of a = "))
b = float(input("Enter the value of b = "))
c = float(input("Enter the value of c = "))

if a == 0:
  print("invalid quadratic equation: the value of a cannot be Zero")
else:
  discriminant = (b**2) - (4*a*c)

  #Condition for real and distinct roots
  if discriminant > 0:
    root1 = (-b + math.sqrt(discriminant)) / (2*a)
    root2 = (-b - math.sqrt(discriminant)) / (2*a)
    print("The roots are real and distinct")
    print(f"root1 = {root1:.2f}")
    print(f"root2 = {root2:.2f}")

  #Conditions for real and equal roots
  elif discriminant == 0:
    root1 = -b / (2*a)
    print("The roots are real and equal")
    print(f"Root 1 = Root 2 = {root1:.2f}")

  else:
    root1 = (-b + cmath.sqrt(discriminant)) / (2*a)
    root2 = (-b - cmath.sqrt(discriminant)) / (2*a)
    print("Roots are complex and imaginary.")
    print(f"Root 1 = {root1.real:.2f} + {root1.imag:.2f}i")
    print(f"Root 2 = {root2.real:.2f} - {abs(root2.imag):.2f}i")
```

```python
#Write a program to read marks from the keyboard and display the equivalent grade according to the table: 80100 Distinction, 60-79 First Class, 40-59 Second Class, below 40 Fail (using if-elif-else).

marks = float(input("Enter the marks obtained: "))
if marks < 0 or marks > 100:
    print("Invalid input! Marks should be between 0 and 100.")
else:
    if marks >= 80:
        grade = "Distinction"
    elif marks >= 60:
        grade = "First Class"
    elif marks >= 40:
        grade = "Second Class"
    else:
        grade = "Fail"
    print(f"Your Grade is: {grade}")
```

```python
#Write a program to read a number from 1 to 7 and print the corresponding day, Sunday to Saturday (using if-elifelse)

day_number = int(input("Enter a number from 1 to 7: "))
if day_number == 1:
    day_name = "Sunday"
elif day_number == 2:
    day_name = "Monday"
elif day_number == 3:
    day_name = "Tuesday"
elif day_number == 4:
    day_name = "Wednesday"
elif day_number == 5:
    day_name = "Thursday"
elif day_number == 6:
    day_name = "Friday"
elif day_number == 7:
    day_name = "Saturday"
else:
    day_name = "Invalid input! Please enter a number strictly between 1 and 7."
print(day_name)
```
